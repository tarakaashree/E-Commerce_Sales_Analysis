```python
#Phase 1: Project Objective
# Global Superstore Sales Analysis

## Objective

#Analyze sales performance, profitability,customer behavior, product trends,and business opportunities.
```

```python
#Phase 2: Import Libraries and Load the DataSet
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
store=pd.read_csv("/content/superstore.csv", on_bad_lines='skip', engine='python')
```

```python
#Phase 3: Data Understanding
print(store.shape)
print(store.ndim)
print(store.info)
print(store.describe())
print(store.head())
print(store.tail())
print(store.columns)
print(store.dtypes)
```

```python
#Phase 4: Data Cleaning
store.isnull() #-->shows True for records that contain null value and False for records that do not contain null value
print(store.isnull().sum()) #--> there is no null value in any column
store.duplicated() #-->shows True for records that contain duplicate value and False for records that do not contain duplicate value
print(store.duplicated().sum()) #--> there is no duplicate value in any column
```

```python
#Phase 5: Feature Engineering
print(store.columns)
#Segregate Order.Date into Year, Month, Date, Day
store['Order.Date']=pd.to_datetime(store['Order.Date'])
store['Year']= store['Order.Date'].dt.year
store['Month']= store['Order.Date'].dt.month
store['Date']= store['Order.Date'].dt.date
store['Day']= store['Order.Date'].dt.day

#Split Customer Name into their First Name and Last Name
store[['First.Name', 'Last.Name']] = store['Customer.Name'].str.split(' ',n=1, expand=True)
store['First.Name'] #--> Displays the First Name of Customers that we have created
store['Last.Name'] #--> Displays the Last Name of Customers that we have created

#Joining City, State and Country into single column as Place
store['Place']= store['City']+ ',' +(store['State'])+ ',' +(store['Country'])
store['Place']

#Create a column Duration to know how many days it take to Ship the order to Customer
store['Ship.Date']=pd.to_datetime(store['Ship.Date'])
store['Duration'] =  store['Ship.Date'] - store['Order.Date']
store['Duration']
```

```python
#Phase 6: Business Kpi

#Total Profit
store['Total.Profit']= store['Profit'].sum(axis=0)
store['Total.Profit']

#Total Order
store['Total.Order']= store['Order.ID'].count()
store['Total.Order']

#Calculate the Profit Margin. Formula= (profit/sales)*100
store['Margin']= (store['Profit']/store['Sales'])*100
store['Margin']

#Total order per Customer
store['Total.Order']= store.groupby('Customer.Name')['Order.ID'].transform('count')
store['Total.Order']

#Average Order Value Per Customer
store['Order_value_per_Customer']= store.groupby('Customer.Name')['Sales'].transform('mean')
store['Order_value_per_Customer']

```

```python
#Phase 7: EDA

#1.Sales Analysis
#Which year generated the highest sales?
Highest_Sales=store.groupby('Year')['Sales'].max().head(1)
Highest_Sales

#Which month generated the lowest sales?
Lowest_Sales=store.groupby('Month')['Sales'].min().head(1)
Lowest_Sales

#Is sales increasing over time?
sns.lineplot(data=store,x='Year',y='Sales')
sns.set_style('whitegrid') #-->Sales is not Increasing over Time

#2.Category Analysis
#Which category sells the most?
store['Category'].value_counts().head(1)

#Which category generates the most profit?
x=store.groupby('Category')['Profit'].sum()
max_category=x.idxmax()
max=x.max()
print(max_category,max)

#Are sales and profit always related?
col=['Sales','Profit']
sns.heatmap(data=store[col].corr(),annot=True, cmap='RdBu')#--> They are moderately correlated

#3.Customer Analysis
#Who are the top 10 customers?
store['Customer.Name'].value_counts().head(10)

#Which segment spends the most?
x=store.groupby('Segment')['Sales'].sum()
max_Segment=x.idxmax()
max_Sales=x.max()
print(max_Segment,max_Sales)

#4.Geographic Analysis
#Best country?(Sales is more)
store.groupby('Country')['Sales'].sum()
max_country=x.idxmax()
max_sa=x.max()
print(max_country,max_sa)

#Worst country?(Sales is less)
store.groupby('Country')['Sales'].sum()
min_country=x.idxmin()
min_sa=x.min()
print(min_country,min_sa)

#5.Profit Analysis
#Which products make losses?
profit=store.groupby('Product.Name')['Profit'].sum()
loss=profit.where(profit<0)
print(loss)
```

```python
#Phase 8: Visualisation
#How has total sales changed over time?
sns.stripplot(data=store,x='Year',y='Sales')#The total sales is decreasing over years
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Year vs Sales')
```

```python
#How does discount percentage affect profit?
columns=['Discount','Profit']
sns.heatmap(data=store[columns].corr(),annot=True,cmap='Blues') # Discount and Profit are negatively correlated to one another. As the Discount increases the Profit decreases by 0.32
```

```python
#Who are the top-performing customers in terms of revenue and profit contribution?
sns.lineplot(data=store,x='Customer.Name',y='Sales',hue='Profit')
plt.xlabel('Customer Name')
plt.ylabel('Sales')
plt.title('Customer Name vs Sales vs Profit')
```
